```python
import pickle
import numpy as np
```

```python
with open("model.pkl", "rb") as f:
    model = pickle.load(f)
with open("scaler.pkl", "rb") as f:
    scaler = pickle.load(f)
```

```python
print("Enter customer details:")

CreditScore = int(input("Credit Score: "))
Age = int(input("Age: "))
Tenure = int(input("Tenure (years): "))
Balance = float(input("Balance: "))
NumOfProducts = int(input("Number of Products: "))
HasCrCard = int(input("Has Credit Card (1 = Yes, 0 = No): "))
IsActiveMember = int(input("Is Active Member (1 = Yes, 0 = No): "))
EstimatedSalary = float(input("Estimated Salary: "))
Satisfaction_Score = int(input("Satisfaction Score (1–5): "))
Point_Earned = int(input("Points Earned: "))

# Geography
geo = input("Geography (Germany/Spain/France): ").lower()
Geography_Germany = 1 if geo == "germany" else 0
Geography_Spain = 1 if geo == "spain" else 0

# Gender
gender = input("Gender (Male/Female): ").lower()
Gender_Male = 1 if gender == "male" else 0

# Card Type
card = input("Card Type (GOLD/PLATINUM/SILVER): ").lower()
Card_Type_GOLD = 1 if card == "gold" else 0
Card_Type_PLATINUM = 1 if card == "platinum" else 0
Card_Type_SILVER = 1 if card == "silver" else 0
```

```python
user_data = np.array([[
    CreditScore,
    Age,
    Tenure,
    Balance,
    NumOfProducts,
    HasCrCard,
    IsActiveMember,
    EstimatedSalary,
    Satisfaction_Score,
    Point_Earned,
    Geography_Germany,
    Geography_Spain,
    Gender_Male,
    Card_Type_GOLD,
    Card_Type_PLATINUM,
    Card_Type_SILVER
]])
```

```python
user_data_scaled = scaler.transform(user_data)

prediction = model.predict(user_data_scaled)
probability = model.predict_proba(user_data_scaled)[0][1]
```

```python
if prediction[0] == 1:
    print(f"⚠️ Customer is likely to CHURN (Probability of Churning: {probability:.2f})")
else:
    print(f"✅ Customer is likely to STAY (Probability of Churning: {probability:.2f})")
```
